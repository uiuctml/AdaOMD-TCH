```python
import pickle
import matplotlib.pyplot as plt
```

```python
def get_table(problem, seeds, methods):
    for method in methods:
        hv = []
        for seed in seeds:
            path = f'discrete/{problem}/{method}/seed_{seed}/res.pickle'
            with open(path, 'rb') as f:
                res = pickle.load(f)
            hv.append(res['hv_arr'][-1])
        avg = sum(hv)/len(hv)
        std = (sum([(x-avg)**2 for x in hv])/len(hv))**0.5
        print(f"{method}: {avg:.3f}±{std:.3f}")
```

```python
methods = ['agg_ls', 'agg_tche', 'agg_softtche', 'agg_omdgdtche', 'agg_gomdgdtche', 'agg_omdegtche', 'agg_gomdegtche', 'excessmtl', 'epo', 'ferero', 'gm_mgda', 'gm_crmogm', 'gm_moco', 'pmtl']
seeds = [0, 19, 42]
```

```python
get_table('F1', seeds, methods)
```
